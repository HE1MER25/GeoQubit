# 1. Classical Bits and Qubits

A classical bit has one of two values:

- `0`
- `1`

A quantum bit, or **qubit**, has two computational-basis states:
$$
|0\rangle
$$

and
$$
|1\rangle.
$$

A general single-qubit state can be written as

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle,
$$
where \(\alpha\) and \(\beta\) are called **probability amplitudes** and

$$
|\alpha|^2 + |\beta|^2 = 1.
$$

For now, the most important idea is:

> A quantum state contains amplitudes. When we measure the qubit, we obtain a classical outcome such as `0` or `1`.

We will see this experimentally rather than trying to learn all of the mathematics at once.

# 5. Visualising a Quantum State

Circuit diagrams show **what operations we applied**.

State visualisations help us understand **what state those operations created**.

For a single qubit, the Bloch sphere is particularly useful.


````python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

qc = QuantumCircuit(1)
qc.h(0)

state = Statevector.from_instruction(qc)

print(state)
plot_bloch_multivector(state)
````

### Important distinction

A **statevector** describes the quantum state before measurement.

# 6. Measurement

To obtain classical information from a quantum circuit, we measure the qubit.

Let's begin with a qubit that is still in its initial state.


````python
qc = QuantumCircuit(1, 1)

qc.measure(0, 0)

qc.draw("mpl")
````

`QuantumCircuit(1, 1)` gives us:

- one qubit; and
- one classical bit.

`qc.measure(0, 0)` measures qubit 0 and stores the result in classical bit 0.


# 7. Running Your First Simulation

We can run the circuit many times on a simulator.

Each repetition is called a **shot**.


````python
result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

print(counts)
````

Because the qubit began in `|0⟩` and we did not change it, the result should be `0` every time in this ideal simulation.

Now let's create a superposition and measure it.


````python
qc = QuantumCircuit(1, 1)

qc.h(0)
qc.measure(0, 0)

result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

print(counts)
````

You should see approximately half `0` and half `1`.

This does **not** mean that one measurement returned both `0` and `1`.

It means that repeated preparations and measurements produced the two possible outcomes with approximately equal frequency.


# 8. Counts and Histograms

Dictionaries of counts are useful, but a histogram makes the distribution easier to see.


````python
from qiskit.visualization import plot_histogram

plot_histogram(counts)
````

# 11. The CNOT / CX Gate

The controlled-X gate acts on two qubits:

```python
qc.cx(control, target)
```

For example:

```python

```

- qubit 0 is the **control**;
- qubit 1 is the **target**.

The target is flipped when the control is in the relevant computational state.


````python
qc = QuantumCircuit(2)

qc.cx(0, 1)

qc.draw("mpl")
````

# 12. Superposition vs Entanglement

Two qubits can each be in superposition without necessarily being entangled.

Let's compare two circuits.

## Circuit A — Independent superpositions


````python
qc_a = QuantumCircuit(2, 2)

qc_a.h(0)
qc_a.h(1)

qc_a.measure([0, 1], [0, 1])

result_a = simulator.run(qc_a, shots=2000).result()

print(result_a.get_counts())
plot_histogram(result_a.get_counts())
````

## Circuit B — Bell-state circuit


````python
qc_b = QuantumCircuit(2, 2)

qc_b.h(0)
qc_b.cx(0, 1)

qc_b.measure([0, 1], [0, 1])

result_b = simulator.run(qc_b, shots=2000).result()

print(result_b.get_counts())
plot_histogram(result_b.get_counts())
````

In the ideal Bell-state experiment, the dominant outcomes correspond to

$$
|00\rangle
$$

and

$$
|11\rangle.
$$

The outcomes are correlated.

Compare that with the first circuit, where all four two-bit outcomes can appear.

This is an important distinction:

> **superposition is not the same thing as entanglement.**

# 16. What Is Transpilation?

The circuit we write is not necessarily the exact sequence of instructions a particular quantum device can execute.

**Transpilation** transforms a circuit into a form suitable for a chosen backend or target.

For now, you only need the idea—not all of the optimisation details.


````python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)

print("Original circuit:")
print(qc)

transpiled_qc = transpile(qc, simulator)

print("\nTranspiled for the simulator:")
print(transpiled_qc)
````

On real hardware, transpilation matters even more because a device has its own:

- supported operations;
- qubit connectivity;
- calibration properties; and
- physical limitations.

Later, you can study transpilation and optimisation in much more detail.


# 17. Simulator → Real Quantum Hardware

So far our workflow has been:

**Circuit → simulator → measurement results**

A real-device workflow adds another important step:

**Circuit → transpilation for a backend → quantum hardware → measurement results**

A useful first hardware experiment is to run the **same Bell-state circuit** on:

1. an ideal/local simulator; and
2. a real quantum backend.

Then compare the histograms.

On real hardware, you may observe outcomes that are absent from an ideal simulation. That leads naturally to the study of **noise, device errors, error mitigation, and eventually quantum error correction**.
